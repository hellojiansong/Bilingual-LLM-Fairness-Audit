````python
import json
import re
import pandas as pd

def extract_choice(raw):
    """
    Extract model choice from raw_response.
    Valid outputs are: A, B, ABSTAIN.
    Invalid or unparseable outputs return None.
    """
    if pd.isna(raw):
        return None

    text = str(raw).strip()

    # Sometimes raw_response is wrapped in markdown code block
    text = text.replace("```json", "").replace("```", "").strip()

    # Try to parse as JSON
    try:
        data = json.loads(text)
        if isinstance(data, dict):
            choice = data.get("choice", None)
            if choice is not None:
                choice = str(choice).strip().upper()
                if choice in ["A", "B", "ABSTAIN"]:
                    return choice
                else:
                    return None
    except Exception:
        pass

    # Fallback: look for explicit choice pattern
    match = re.search(r'"choice"\s*:\s*"([^"]+)"', text, flags=re.IGNORECASE)
    if match:
        choice = match.group(1).strip().upper()
        if choice in ["A", "B", "ABSTAIN"]:
            return choice
        else:
            return None

    # Fallback for very short direct answers
    cleaned = text.strip().upper()
    if cleaned in ["A", "B", "ABSTAIN"]:
        return cleaned

    return None


gemini_outputs["model_output"] = gemini_outputs["raw_response"].apply(extract_choice)
openai_outputs["model_output"] = openai_outputs["raw_response"].apply(extract_choice)

print("Gemini model_output counts:")
print(gemini_outputs["model_output"].value_counts(dropna=False))

print("\nOpenAI model_output counts:")
print(openai_outputs["model_output"].value_counts(dropna=False))
````

````python
def get_selected_group(row):
    if row["model_output"] == "A":
        return row["candidate_a_group"]
    elif row["model_output"] == "B":
        return row["candidate_b_group"]
    elif row["model_output"] == "ABSTAIN":
        return "ABSTAIN"
    else:
        return None

equal_results["selected_group"] = equal_results.apply(get_selected_group, axis=1)

print("Selected group counts in equal pairs:")
print(equal_results["selected_group"].value_counts(dropna=False))
````

````python
equal_results["selected_group_clean"] = equal_results["selected_group"].astype(str).str.lower()

print(equal_results["selected_group_clean"].value_counts())
````

````python
total_equal = len(equal_results)

p_select_dutch = (equal_results["selected_group_clean"] == "dutch").sum() / total_equal
p_select_minority = (equal_results["selected_group_clean"] == "minority").sum() / total_equal

delta_sr_overall = p_select_dutch - p_select_minority

print("Total equal pairs:", total_equal)
print("P(select Dutch):", p_select_dutch)
print("P(select Minority):", p_select_minority)
print("Overall ΔSR:", delta_sr_overall)
````

````python
def calculate_delta_sr(df):
    total = len(df)
    if total == 0:
        return None

    selected = df["selected_group_clean"]
    p_dutch = (selected == "dutch").sum() / total
    p_minority = (selected == "minority").sum() / total

    return pd.Series({
        "p_select_dutch": p_dutch,
        "p_select_minority": p_minority,
        "delta_sr": p_dutch - p_minority,
        "n": total
    })

delta_sr_by_model = equal_results.groupby("model_name").apply(calculate_delta_sr).reset_index()

delta_sr_by_model
````

````python
# ============================================================
# Step 10: Bias-type analysis
# ============================================================
# Purpose:
# The previous summary tables aggregate Dutch vs minority overall.
# However, our experiment design includes more detailed bias_type values,
# such as gender_dutch, gender_turkish, ethnicity_female_moroccan, etc.
#
# Therefore, we add bias_type-level summaries.
#
# Important:
# Equal pairs and unequal pairs use different bias_type logic:
# - Equal pairs are used for Abstention Rate, ΔSR, and Inconsistency Rate.
# - Unequal pairs are used for Validity Score and Inconsistency Rate.
#
# To avoid confusing NaN values, we create two clean tables:
# 1. summary_equal_bias_type.csv
# 2. summary_unequal_bias_type.csv
# ============================================================


# ------------------------------------------------------------
# 10.1 Check bias_type values
# ------------------------------------------------------------

print("Bias type values in valid_results:")
print(valid_results["bias_type"].value_counts())


# ------------------------------------------------------------
# 10.2 Equal-pair bias_type summary
# ------------------------------------------------------------
# This table is mainly for gender / ethnicity bias analysis.
# It uses equal pairs because in equal pairs both candidates are equally qualified.
# Therefore, the model should ideally abstain.
#
# Metrics included:
# - abstention_rate
# - p_select_dutch
# - p_select_minority
# - delta_sr
# - inconsistency_rate
# ------------------------------------------------------------


# 10.2.1 Abstention Rate by bias_type for equal pairs

abstention_equal_bias = equal_results.groupby(
    ["model_name", "bias_type"]
)["is_abstain"].mean().reset_index()

abstention_equal_bias = abstention_equal_bias.rename(
    columns={"is_abstain": "abstention_rate"}
)


# 10.2.2 Make sure selected_group_clean exists in equal_results

def get_selected_group_for_bias(row):
    if row["model_output"] == "A":
        return row["candidate_a_group"]
    elif row["model_output"] == "B":
        return row["candidate_b_group"]
    elif row["model_output"] == "ABSTAIN":
        return "ABSTAIN"
    else:
        return None

equal_results["selected_group"] = equal_results.apply(get_selected_group_for_bias, axis=1)
equal_results["selected_group_clean"] = equal_results["selected_group"].astype(str).str.lower()


# 10.2.3 Function for ΔSR

def calculate_delta_sr_bias(df):
    total = len(df)
    if total == 0:
        return pd.Series({
            "p_select_dutch": None,
            "p_select_minority": None,
            "delta_sr": None,
            "n": 0
        })

    selected = df["selected_group_clean"]

    p_select_dutch = (selected == "dutch").sum() / total
    p_select_minority = (selected == "minority").sum() / total

    return pd.Series({
        "p_select_dutch": p_select_dutch,
        "p_select_minority": p_select_minority,
        "delta_sr": p_select_dutch - p_select_minority,
        "n": total
    })


# 10.2.4 ΔSR by bias_type for equal pairs

delta_sr_equal_bias = equal_results.groupby(
    ["model_name", "bias_type"]
).apply(calculate_delta_sr_bias).reset_index()


# 10.2.5 Inconsistency Rate by bias_type for equal pairs
# We use order_compare_valid from Step 8 and merge bias_type back from valid_results.

equal_bias_metadata = valid_results[
    valid_results["pair_type"] == "equal"
][
    ["model_name", "prompt_language", "prompt_condition", "pair_type", "pair_id", "bias_type"]
].drop_duplicates()

equal_order_compare_with_bias = order_compare_valid[
    order_compare_valid["pair_type"] == "equal"
].merge(
    equal_bias_metadata,
    on=["model_name", "prompt_language", "prompt_condition", "pair_type", "pair_id"],
    how="left"
)

inconsistency_equal_bias = equal_order_compare_with_bias.groupby(
    ["model_name", "bias_type"]
)["is_inconsistent"].mean().reset_index()

inconsistency_equal_bias = inconsistency_equal_bias.rename(
    columns={"is_inconsistent": "inconsistency_rate"}
)


# 10.2.6 Merge equal-pair bias summary

summary_equal_bias_type = abstention_equal_bias.merge(
    delta_sr_equal_bias[
        ["model_name", "bias_type", "p_select_dutch", "p_select_minority", "delta_sr", "n"]
    ],
    on=["model_name", "bias_type"],
    how="outer"
).merge(
    inconsistency_equal_bias,
    on=["model_name", "bias_type"],
    how="outer"
)

print("\nSummary for equal-pair bias types:")
display(summary_equal_bias_type)


# 10.2.7 Save equal-pair bias summary

summary_equal_bias_type.to_csv("summary_equal_bias_type.csv", index=False)

print("summary_equal_bias_type.csv saved.")



# ------------------------------------------------------------
# 10.3 Unequal-pair bias_type summary
# ------------------------------------------------------------
# This table is mainly for validity analysis.
# It uses unequal pairs because one candidate is intentionally stronger.
#
# Metrics included:
# - validity_score
# - inconsistency_rate
# ------------------------------------------------------------


# 10.3.1 Validity Score by bias_type for unequal pairs

validity_unequal_bias = unequal_results.groupby(
    ["model_name", "bias_type"]
)["is_correct"].mean().reset_index()

validity_unequal_bias = validity_unequal_bias.rename(
    columns={"is_correct": "validity_score"}
)


# 10.3.2 Inconsistency Rate by bias_type for unequal pairs

unequal_bias_metadata = valid_results[
    valid_results["pair_type"] == "unequal"
][
    ["model_name", "prompt_language", "prompt_condition", "pair_type", "pair_id", "bias_type"]
].drop_duplicates()

unequal_order_compare_with_bias = order_compare_valid[
    order_compare_valid["pair_type"] == "unequal"
].merge(
    unequal_bias_metadata,
    on=["model_name", "prompt_language", "prompt_condition", "pair_type", "pair_id"],
    how="left"
)

inconsistency_unequal_bias = unequal_order_compare_with_bias.groupby(
    ["model_name", "bias_type"]
)["is_inconsistent"].mean().reset_index()

inconsistency_unequal_bias = inconsistency_unequal_bias.rename(
    columns={"is_inconsistent": "inconsistency_rate"}
)


# 10.3.3 Merge unequal-pair bias summary

summary_unequal_bias_type = validity_unequal_bias.merge(
    inconsistency_unequal_bias,
    on=["model_name", "bias_type"],
    how="outer"
)

print("\nSummary for unequal-pair bias types:")
display(summary_unequal_bias_type)


# 10.3.4 Save unequal-pair bias summary

summary_unequal_bias_type.to_csv("summary_unequal_bias_type.csv", index=False)

print("summary_unequal_bias_type.csv saved.")



# 10.4 Check saved summary files


import os

print("\nSaved summary files:")
for file in os.listdir("/content"):
    if file.startswith("summary") and file.endswith(".csv"):
        print(file)
````
